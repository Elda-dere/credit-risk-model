# Exploratory Data Analysis - Credit Risk Model

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/germany.csv')
print(f'Dataset shape: {df.shape}')
df.head()
```

```python
print('Missing values:')
print(df.isnull().sum())
```

```python
positive_amounts = df[df['Amount'] > 0]['Amount']

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.hist(positive_amounts, bins=50, color='steelblue')
plt.title('Transaction Amounts')
plt.subplot(1,2,2)
plt.hist(np.log1p(positive_amounts), bins=50, color='green')
plt.title('Log Transformed')
plt.show()

print(f'Skewness: {positive_amounts.skew():.2f}')
```

## TOP 5 INSIGHTS

1. No missing values - Clean dataset
2. Transaction amounts are highly skewed - Use log transformation
3. Low fraud rate (~1.5%)
4. RFM segmentation can create proxy default
5. Categorical features need encoding (WoE recommended)
