```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported!")
```

```python
df = pd.read_csv('../data/raw/germany.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

## 1. Data Overview

```python
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
print(f"\nColumn names:\n{df.columns.tolist()}")
```

## 2. Missing Values

```python
missing = df.isnull().sum()
missing = missing[missing > 0]
if len(missing) > 0:
    print("Columns with missing values:")
    print(missing)
else:
    print("✅ NO MISSING VALUES FOUND!")
```

## 3. Distribution of Transaction Amounts

```python
positive_amounts = df[df['Amount'] > 0]['Amount']

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(positive_amounts, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
axes[0].set_title('Transaction Amount Distribution')
axes[0].set_xlabel('Amount')

axes[1].hist(np.log1p(positive_amounts), bins=50, color='forestgreen', edgecolor='black', alpha=0.7)
axes[1].set_title('Log-Transformed Amount')
axes[1].set_xlabel('Log(Amount + 1)')

plt.tight_layout()
plt.show()

print(f"Skewness: {positive_amounts.skew():.2f}")
print(f"Skewness after log: {np.log1p(positive_amounts).skew():.2f}")
```

## 4. Fraud Analysis

```python
fraud_counts = df['FraudResult'].value_counts()
fraud_pct = df['FraudResult'].value_counts(normalize=True) * 100

plt.figure(figsize=(6, 4))
plt.bar(fraud_counts.index, fraud_counts.values, color=['green', 'red'], alpha=0.7)
plt.title('Fraud Distribution')
plt.xlabel('FraudResult (0=No, 1=Yes)')
plt.ylabel('Count')
plt.show()

print(f"Fraud Rate: {fraud_pct[1]:.2f}%")
```

## 5. TOP 5 KEY INSIGHTS

### Insight 1: Excellent Data Quality ✅
- No missing values across all columns
- No imputation needed

### Insight 2: Skewed Transaction Distribution 📊
- Amounts are highly right-skewed (skewness > 10)
- Log transformation normalizes distribution
- Recommendation: Use log1p transformation

### Insight 3: Low Fraud Rate 🔍
- Only ~1.5% of transactions are fraudulent
- Highly imbalanced - consider class weights

### Insight 4: Categorical Feature Variation 🏷️
- ProductCategory has multiple distinct values
- ChannelId has multiple distinct values
- Recommendation: Use frequency encoding or WoE

### Insight 5: RFM Segmentation
- Can create proxy default using RFM
- Provides reasonable target distribution

### Modeling Recommendations
1. Apply log1p transformation to Amount/Value
2. Encode categorical features (WoE recommended)
3. Include FraudResult as predictive feature
4. Create customer-level RFM aggregations
5. Consider ensemble methods for non-linear relationships
