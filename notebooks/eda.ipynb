```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully!")
```

```python
# Load data
df = pd.read_csv("../data/raw/germany.csv")
print(f"Dataset shape: {df.shape}")
df.head()
```

## 1. Data Overview

```python
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
print(f"\nColumn names:\n{df.columns.tolist()}")
print(f"\nData types:\n{df.dtypes}")
```

## 2. Summary Statistics

```python
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"Numerical columns: {numerical_cols}\n")
df[numerical_cols].describe()
```

## 3. Distribution of Numerical Features

```python
positive_amounts = df[df['Amount'] > 0]['Amount']

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(positive_amounts, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
axes[0].set_title('Transaction Amount Distribution')
axes[0].set_xlabel('Amount')

axes[1].hist(np.log1p(positive_amounts), bins=50, color='forestgreen', edgecolor='black', alpha=0.7)
axes[1].set_title('Log-Transformed Amount')
axes[1].set_xlabel('Log(Amount + 1)')

plt.tight_layout()
plt.show()

print(f"Skewness: {positive_amounts.skew():.2f}")
print(f"Skewness after log: {np.log1p(positive_amounts).skew():.2f}")
```

## 4. Distribution of Categorical Features

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

top_products = df['ProductCategory'].value_counts().head(10)
axes[0].barh(range(len(top_products)), top_products.values, color='teal', alpha=0.7)
axes[0].set_yticks(range(len(top_products)))
axes[0].set_yticklabels(top_products.index)
axes[0].set_title('Top 10 Product Categories')

top_channels = df['ChannelId'].value_counts().head(10)
axes[1].barh(range(len(top_channels)), top_channels.values, color='coral', alpha=0.7)
axes[1].set_yticks(range(len(top_channels)))
axes[1].set_yticklabels(top_channels.index)
axes[1].set_title('Top 10 Channels')

plt.tight_layout()
plt.show()
```

## 5. Correlation Analysis

```python
plt.figure(figsize=(8, 6))
corr_matrix = df[numerical_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## 6. Missing Values

```python
missing = df.isnull().sum()
missing = missing[missing > 0]
if len(missing) > 0:
    print("Columns with missing values:")
    print(missing)
else:
    print("✅ NO MISSING VALUES FOUND!")
```

## 7. Outlier Detection

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

amount_capped = positive_amounts[positive_amounts < positive_amounts.quantile(0.99)]
axes[0].boxplot(amount_capped)
axes[0].set_title('Transaction Amount (99th percentile capped)')

value_capped = df['Value'][df['Value'] < df['Value'].quantile(0.99)]
axes[1].boxplot(value_capped)
axes[1].set_title('Absolute Value (99th percentile capped)')

plt.tight_layout()
plt.show()
```

## 8. Fraud Analysis

```python
fraud_counts = df['FraudResult'].value_counts()
fraud_pct = df['FraudResult'].value_counts(normalize=True) * 100

plt.figure(figsize=(6, 4))
plt.bar(fraud_counts.index, fraud_counts.values, color=['green', 'red'], alpha=0.7)
plt.title('Fraud Distribution')
plt.xlabel('FraudResult (0=No, 1=Yes)')
plt.ylabel('Count')
plt.show()

print(f"Fraud Rate: {fraud_pct[1]:.2f}%")
```

## 9. Top 5 Key Insights

### Insight 1: Excellent Data Quality ✅
- No missing values across all columns
- No imputation needed

### Insight 2: Skewed Transaction Distribution 📊
- Amounts are highly right-skewed (skewness > 10)
- Log transformation normalizes distribution
- Recommendation: Use log1p transformation

### Insight 3: Low Fraud Rate 🔍
- Only ~1.5% of transactions are fraudulent
- Highly imbalanced - consider class weights

### Insight 4: Categorical Feature Variation 🏷️
- ProductCategory has 12 distinct values
- ChannelId has 5 distinct values
- Recommendation: Use frequency encoding or WoE

### Insight 5: RFM Segmentation
- Can create proxy default using RFM
- Provides reasonable target distribution

### Modeling Recommendations
1. Apply log1p transformation to Amount/Value
2. Encode categorical features (WoE recommended)
3. Include FraudResult as predictive feature
4. Create customer-level RFM aggregations
5. Consider ensemble methods for non-linear relationships
